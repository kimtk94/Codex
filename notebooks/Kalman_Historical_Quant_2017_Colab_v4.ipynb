```python
from google.colab import drive
import shutil
import subprocess
from pathlib import Path

drive.mount("/content/drive", force_remount=False)

repo = Path("/content/Codex")
if repo.exists():
    shutil.rmtree(repo)

subprocess.run(
    [
        "git",
        "clone",
        "--depth",
        "1",
        "https://github.com/kimtk94/Codex.git",
        str(repo),
    ],
    check=True,
)

gateway = next(repo.glob("kalman-*-gateway"))\nrunner = gateway / "scripts" / "colab_historical_quant.py"

# Self-test first. Historical run starts only if this passes.
subprocess.run(
    ["python", str(runner), "--self-test"],
    check=True,
)

subprocess.run(
    [
        "python",
        str(runner),
        "--drive-root",
        "/content/drive/MyDrive",
        "--start-date",
        "2017-01-01",
    ],
    check=True,
)
```
