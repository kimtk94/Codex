```python
from google.colab import drive
import shutil
import subprocess
from pathlib import Path

drive.mount('/content/drive', force_remount=False)
repo = Path('/content/Codex')
if repo.exists():
    shutil.rmtree(repo)
subprocess.run(
    ['git', 'clone', '--depth', '1', '--branch', 'main', 'https://github.com/kimtk94/Codex.git', str(repo)],
    check=True,
)
gateway = next(repo.glob('kalman-*-gateway'))
runner = gateway / 'scripts' / 'colab_resume_historical.py'
assert runner.exists(), runner
subprocess.run(
    [
        'python',
        str(runner),
        '--drive-root',
        '/content/drive/MyDrive',
        '--run-tag',
        '20260913_042850',
    ],
    check=True,
)
```
