```python
# STAGE 2-C5C4 — ONE-CELL COLAB LAUNCHER
# Runs the verified multi-step C5C4 notebook from your mounted Google Drive.

import json
from pathlib import Path
from google.colab import drive

drive.mount("/content/drive")

root = Path("/content/drive/MyDrive")
name = "STAGE2_C5C4_Corrected_SuSiE_Colab_copy.ipynb"

matches = list(root.rglob(name))
if not matches:
    raise FileNotFoundError(f"{name} not found under MyDrive")

src = matches[0]
print("SOURCE NOTEBOOK:", src)

nb = json.loads(src.read_text(encoding="utf-8"))
code_cells = [c for c in nb.get("cells", []) if c.get("cell_type") == "code"]

print("CODE CELLS TO RUN:", len(code_cells))
print("=" * 60)

g = globals()

for i, c in enumerate(code_cells, start=1):
    source = c.get("source", "")
    if isinstance(source, list):
        source = "".join(source)

    print(f"\n===== INTERNAL STEP {i}/{len(code_cells)} =====")
    exec(compile(source, f"<C5C4-step-{i}>", "exec"), g, g)

print("\n" + "=" * 60)
print("STAGE 2-C5C4 ONE-CELL PIPELINE COMPLETE")
```
