# Multi-Organ Aging — Drive-first Bootstrap
Mount Drive, checkout the research branch, install dependencies, run tests, and snapshot code into MASTER_DEGREE/MULTI_ORGAN_AGING/02_CODE.

```python
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
DRIVE_ROOT = Path('/content/drive/MyDrive/MASTER_DEGREE/MULTI_ORGAN_AGING')
for name in ['00_MASTER','01_INPUT','02_CODE','03_RESULTS','04_REPORTS','99_ARCHIVE']:
    (DRIVE_ROOT / name).mkdir(parents=True, exist_ok=True)
print('Drive root:', DRIVE_ROOT)
```

```python
import subprocess
REPO = Path('/content/Codex')
BRANCH = 'research/multi-organ-aging-20260927'
if not (REPO / '.git').exists():
    subprocess.run(['git','clone','--branch',BRANCH,'--single-branch','https://github.com/kimtk94/Codex.git',str(REPO)], check=True)
else:
    subprocess.run(['git','-C',str(REPO),'fetch','origin',BRANCH], check=True)
    subprocess.run(['git','-C',str(REPO),'checkout',BRANCH], check=True)
    subprocess.run(['git','-C',str(REPO),'pull','--ff-only','origin',BRANCH], check=True)
BASE = REPO / 'master_degree' / 'multi_organ_aging'
print('Code:', BASE)
```

```python
import sys, subprocess
subprocess.run([sys.executable,'-m','pip','install','-q','-r',str(BASE/'requirements.txt')], check=True)
subprocess.run([sys.executable,'-m','compileall',str(BASE/'src')], check=True)
subprocess.run([sys.executable,'-m','pytest','-q',str(BASE/'tests')], check=True)
```

```python
import shutil, datetime
stamp = datetime.datetime.now().strftime('%Y%m%d_%H%M%S')
snapshot = DRIVE_ROOT / '02_CODE' / f'git_snapshot_{stamp}'
shutil.copytree(BASE, snapshot)
print('Code snapshot:', snapshot)
```
